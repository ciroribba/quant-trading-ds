```python
import yfinance as yf

print(yf.__version__)
```

```python
btc = yf.download(
    "BTC-USD",
    start="2018-01-01",
    auto_adjust=False
)
```

```python
btc.columns = btc.columns.get_level_values(0)
```

```python
btc["Close"].plot(
    figsize=(14, 6),
    title="Bitcoin - Precio de cierre"
)

plt.xlabel("Fecha")
plt.ylabel("Precio USD")
plt.show()
```
